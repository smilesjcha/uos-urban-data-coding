### Functions Are Objects

```python
states = ['   Alabama ', 'Georgia!', 'Georgia', 'georgia', 'FlOrIda',
          'south   carolina##', 'West virginia?']
```

```python
import re

def clean_strings(strings):
    result = []
    for value in strings:
        value = value.strip()
        value = re.sub('[!#?]', '', value)
        value = value.title()
        result.append(value)
    return result
```

```python
import re

states = ['   Alabama ', 'Georgia!', 'Georgia', 'georgia', 'FlOrIda',
          'south   carolina##', 'West virginia?']

# 정규표현식 패턴
pattern = r'[A-Za-z\s]*[aA-zZ][A-Za-z\s]*'

# 주를 추출
us_states = [re.search(pattern, state).group().strip() for state in states]

# 중복 제거
us_states = list(set(us_states))

print(us_states)
```

```python
import re

states = ['   Alabama ', 'Georgia!', 'Georgia', 'georgia', 'FlOrIda',
          'south   carolina##', 'West virginia?']

# 정규표현식 패턴
pattern = r'[A-Za-z\s]*[aA-zZ][A-Za-z\s]*'

# 주 이름을 추출하고 변환
us_states = [re.search(pattern, state).group().strip() for state in states]

# 주 이름을 원하는 형식으로 변환
us_states = [state.capitalize() for state in us_states]

print(us_states)
```

```python
clean_strings(states)
```

```python
def remove_punctuation(value):
    return re.sub('[!#?]', '', value)

clean_ops = [str.strip, remove_punctuation, str.title]

def clean_strings(strings, ops):
    result = []
    for value in strings:
        for function in ops:
            value = function(value)
        result.append(value)
    return result
```

```python
clean_strings(states, clean_ops)
```

```python
for x in map(remove_punctuation, states):
    print(x)
```

```python
def squares(n=10):
    print('Generating squares from 1 to {0}'.format(n ** 2))
    for i in range(1, n + 1):
        yield i ** 2
```
